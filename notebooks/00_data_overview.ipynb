```python
from pathlib import Path

import pandas as pd

DATA = Path("../data/raw")

raw = pd.read_parquet(DATA / "prices_se3_raw.parquet")
hourly = pd.read_parquet(DATA / "prices_se3_hourly.parquet")
weather = pd.read_parquet(DATA / "weather_stockholm_hourly.parquet")

for name, df in [("raw prices", raw), ("hourly prices", hourly), ("weather", weather)]:
    print(name, df.shape, "missing values:", df.isna().sum().sum())
```

```python
print("duplicate start times:", raw["time_start"].duplicated().sum())
print(hourly["time_utc"].diff().value_counts())

full = pd.date_range(weather["time_utc"].min(), weather["time_utc"].max(), freq="h")
print("missing weather hours:", len(full) - len(weather))
```

## Findings

I checked the three data files before using them for modelling.

**Size and completeness**
- Raw prices: 60,311 rows. Hourly prices: 34,247 rows. Weather: 34,224 rows.
- None of the files have missing values.
- The hourly prices have no gaps. Every row is exactly one hour after the one before.
- The weather data also has no missing hours. It ends on 26 September because the archive is a few days behind.

**Change in price resolution**
- Before 1 October 2025 the prices are hourly. After that they come in 15-minute steps.
- This is why the raw file has more rows than the hourly file.
- I averaged the 15-minute prices into hourly prices so the whole dataset has the same format.

**Problem in the source data**
- Three rows in the raw file have the wrong end time.
- They are on 2023-10-29, 2024-10-27 and 2025-10-26. These are the nights when Sweden goes back to winter time.
- On those nights the clock shows 02:00 twice. The source gives the wrong end time for one of these hours, so it looks like the interval is 75 or 120 minutes long.
- The start times are correct, and there are no duplicate start times.

**Why this does not affect the project**
- My code only uses the start time of each price. The end time is never used.
- The check on the hourly file shows every hour follows the previous one exactly, so no hours are lost or counted twice.

**Weather values**
- Temperature is between -18.2 and 30.0 °C, wind speed up to 42 km/h, and solar radiation from 0 to 834 W/m². All of these look normal for Stockholm.
